```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# --------------------------------------------------
# 1. Create the data
# --------------------------------------------------

data = {
    "salary": [1.7, 2.4, 2.3, 3.1, 3.7, 4.2, 4.4, 6.1, 5.4, 5.7, 6.4, 6.2],
    "experience": [1.2, 1.5, 1.9, 2.2, 2.4, 2.5, 2.8, 3.1, 3.3, 3.7, 4.2, 4.4]
}

df = pd.DataFrame(data)

x = df["experience"].values
y = df["salary"].values


# --------------------------------------------------
# 2. Calculate MSE
# --------------------------------------------------

def calculate_mse(beta, b):
    y_pred = beta * x + b
    mse = np.mean((y - y_pred) ** 2)
    return mse


# --------------------------------------------------
# 3. Plot original data
# --------------------------------------------------

plt.scatter(x, y)
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Salary vs Experience")
plt.show()


# --------------------------------------------------
# (a) Check beta = 0.1, 1.5, 0.8
#     with fixed intercept b = 1.1
# --------------------------------------------------

betas = [0.1, 1.5, 0.8]
b = 1.1

for beta in betas:
    mse = calculate_mse(beta, b)
    print("Beta =", beta, "MSE =", mse)

    y_pred = beta * x + b

    plt.scatter(x, y)
    plt.plot(x, y_pred)
    plt.xlabel("Experience")
    plt.ylabel("Salary")
    plt.title("Beta = " + str(beta) + ", MSE = " + str(round(mse, 4)))
    plt.show()


# --------------------------------------------------
# (b) Try beta from 0.01 to 1.5
#     with increment 0.01
#     b = 1.1
# --------------------------------------------------

beta_values = np.arange(0.01, 1.51, 0.01)
mse_values = []

for beta in beta_values:
    mse = calculate_mse(beta, 1.1)
    mse_values.append(mse)

plt.plot(beta_values, mse_values)
plt.xlabel("Beta")
plt.ylabel("Mean Squared Error")
plt.title("Beta vs MSE (Intercept = 1.1)")
plt.show()

best_beta = beta_values[np.argmin(mse_values)]
best_mse = min(mse_values)

print("\nBest beta =", best_beta)
print("Minimum MSE =", best_mse)


# --------------------------------------------------
# (c) Try different intercept values
#     from 0 to 1.5
# --------------------------------------------------

intercept_values = np.arange(0, 1.51, 0.01)
mse_intercept = []

# Use the best beta obtained above
for b in intercept_values:
    mse = calculate_mse(best_beta, b)
    mse_intercept.append(mse)

plt.plot(intercept_values, mse_intercept)
plt.xlabel("Intercept")
plt.ylabel("Mean Squared Error")
plt.title("Intercept vs MSE")
plt.show()

best_intercept = intercept_values[np.argmin(mse_intercept)]
best_intercept_mse = min(mse_intercept)

print("\nBest intercept =", best_intercept)
print("Minimum MSE =", best_intercept_mse)


# --------------------------------------------------
# (d) Using Scikit-learn
# --------------------------------------------------

model = LinearRegression()

X = x.reshape(-1, 1)

model.fit(X, y)

beta_sklearn = model.coef_[0]
b_sklearn = model.intercept_

y_pred_sklearn = model.predict(X)

mse_sklearn = mean_squared_error(y, y_pred_sklearn)

print("\nScikit-learn Results")
print("--------------------")
print("Beta (slope) =", beta_sklearn)
print("Intercept =", b_sklearn)
print("MSE =", mse_sklearn)

plt.scatter(x, y)
plt.plot(x, y_pred_sklearn)
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Scikit-learn Linear Regression")
plt.show()
```
